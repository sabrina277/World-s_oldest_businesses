```python

businesses=pd.read_csv("data/businesses.csv")


new_businesses=pd.read_csv("data/new_businesses.csv")

categories=pd.read_csv("data/categories.csv")

countries=pd.read_csv("data/countries.csv")




import os
print(os.getcwd())
```

```python
# What is the oldest business on each continent? Save your answer as a DataFrame
#  called oldest_business_continent with four columns: continent, country, business, and year_founded in any order.
print(businesses.query("country_code == 'FJI'"))
all_businesses = pd.concat([businesses, new_businesses],
                         ignore_index=True)
print(all_businesses.columns)

merged_businesses = pd.merge(
    all_businesses, 
    countries, 
    on='country_code', 
    how='left'
)

oldest_business_continent = merged_businesses.sort_values(by='year_founded', ascending=True).groupby('continent', as_index=False).first()

oldest_business_continent = oldest_business_continent.drop(columns=['category_code','country_code'])

oldest_business_continent
```

```python
#How many countries per continent lack data on the oldest businesses? Does including new_businesses change this? Count the number of countries per continent missing business data, including new_businesses, and store the results in a 
# DataFrame named count_missing with columns for the continent and the count.

# asia 6 countries missing.
# north amercia 4 countries missing.
# south 3 countries missing.

missing_businesses = countries.merge(
    all_businesses,
    on="country_code",
    how="left"
)

# FIX: Groupby on DataFrame, not Series, and count missing businesses per continent
all_missing_businesses = missing_businesses[missing_businesses['business'].isnull()]
print(all_missing_businesses)
count_missing = all_missing_businesses.groupby('continent').size().reset_index(name='count')
count_missing
```

```python
# Which business categories are best suited to last many years, and on what continent are they? 
# Create a DataFrame called oldest_by_continent_category that stores the oldest founding year for each continent and category combination.
#  It should contain three columns: continent, category, and year_founded, in that order.

re_merging = merged_businesses[['year_founded','continent', 'category_code']].merge(
    categories[['category', 'category_code']],
    on='category_code',
    how='left'
)
re_merging
oldest_by_continent_category = re_merging[['year_founded','continent', 'category']]
oldest_by_continent_category
oldest_by_continent_category = oldest_by_continent_category.groupby(['continent','category'],as_index=False).min()
oldest_by_continent_category
```
